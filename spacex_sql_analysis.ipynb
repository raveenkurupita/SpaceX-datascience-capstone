```python
%load_ext sql

# Load data into SQLite
import pandas as pd
import sqlite3
con = sqlite3.connect("spacex.db")
df = pd.read_csv("https://github.com/raveenkurupita/SpaceX-datascience-capstone/raw/main/dataset_part1.csv")
df.to_sql("SPACEXTBL", con, if_exists='replace', index=False, method="multi")
print("Data loaded!")

# Connect
%sql sqlite:///spacex.db

# TASK 1
print("\nTASK 1: Unique Launch Sites")
%sql SELECT DISTINCT LaunchSite FROM SPACEXTBL
```

```python
# TASK 2: 5 records where launch sites begin with 'CCA'
print("\nTASK 2: Launch Sites starting with CCA")
%sql SELECT * FROM SPACEXTBL WHERE LaunchSite LIKE 'CCA%' LIMIT 5

# TASK 3: Total payload mass for NASA
print("\nTASK 3: Total Payload for NASA")
%sql SELECT SUM(PayloadMass) AS Total_Payload FROM SPACEXTBL WHERE LaunchSite LIKE '%KSC%'

# TASK 4: Average payload for F9 v1.1
print("\nTASK 4: Avg Payload F9 v1.1")
%sql SELECT AVG(PayloadMass) AS Avg_Payload FROM SPACEXTBL WHERE "Booster Version" = 'F9 v1.1'

# TASK 5: First successful landing
print("\nTASK 5: First Success")
%sql SELECT MIN("Unnamed: 0") AS First_Success FROM SPACEXTBL WHERE Class = 1

# TASK 6: Success drone ship payload 4000-6000
print("\nTASK 6: Drone Ship Success 4000-6000")
%sql SELECT "Booster Version" FROM SPACEXTBL WHERE Class = 1 AND PayloadMass BETWEEN 4000 AND 6000

# TASK 7: Total Success vs Failure
print("\nTASK 7: Success vs Failure Count")
%sql SELECT Class, COUNT(*) AS Total FROM SPACEXTBL GROUP BY Class

# TASK 8: Boosters with Max Payload
print("\nTASK 8: Boosters with Max Payload")
%sql SELECT "Booster Version" FROM SPACEXTBL WHERE PayloadMass = (SELECT MAX(PayloadMass) FROM SPACEXTBL)
```
