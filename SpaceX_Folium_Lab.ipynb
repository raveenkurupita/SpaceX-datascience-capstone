```python
# ================== SPACE X FOLIUM LAB ==================

import folium
from folium.plugins import MarkerCluster
import pandas as pd
from math import sin, cos, sqrt, atan2, radians

# 1. Load data
spacex_df = pd.read_csv("https://github.com/raveenkurupita/SpaceX-datascience-capstone/raw/main/spacex_launch_dash.csv")

# Launch site coordinates - manual because not in CSV
launch_site_coords = {
    'CCAFS LC-40': [28.562302, -80.577356],
    'CCAFS SLC-40': [28.563197, -80.576820],
    'KSC LC-39A': [28.573255, -80.646895],
    'VAFB SLC-4E': [34.632834, -120.610745]
}

# 2. Map centered at NASA JSC
nasa_coordinate = [29.559684888503615, -95.0830971930759]
site_map = folium.Map(location=nasa_coordinate, zoom_start=5)
folium.Circle(nasa_coordinate, radius=1000, color='#d35400', fill=True).add_child(folium.Popup('NASA Johnson Space Center')).add_to(site_map)
folium.Marker(nasa_coordinate).add_to(site_map)
site_map
```

```python
# 3. Add all Launch Sites with Blue Rocket Markers
spacex_map = folium.Map(location=[28.563197, -80.576820], zoom_start=4)

for site, coords in launch_site_coords.items():
    folium.Marker(
        location=coords,
        popup=site,
        icon=folium.Icon(color='blue', icon='rocket', prefix='fa')
    ).add_to(spacex_map)

spacex_map
```

```python
# 4. Add colored markers: Green=Success, Red=Failure
marker_cluster = MarkerCluster().add_to(spacex_map)

def get_marker_color(launch_outcome):
    return 'green' if launch_outcome == 1 else 'red'

for index, row in spacex_df.iterrows():
    coords = launch_site_coords[row['Launch Site']]
    folium.Marker(
        location=coords,
        icon=folium.Icon(color=get_marker_color(row['class']))
    ).add_to(marker_cluster)

spacex_map
```

```python
# 5. Distance calculation - FIXED VERSION
def calculate_distance(lat1, lon1, lat2, lon2):
    R = 6373.0
    lat1, lon1, lat2, lon2 = map(radians, [lat1, lon1, lat2, lon2])
    dlon = lon2 - lon1
    dlat = lat2 - lat1
    a = sin(dlat / 2)*2 + cos(lat1) * cos(lat2) * sin(dlon / 2)*2
    a = min(1, max(0, a)) # THIS LINE FIXES THE ERROR
    c = 2 * atan2(sqrt(a), sqrt(1 - a))
    return R * c

ksc_coord = launch_site_coords['KSC LC-39A']
orlando_coord = [28.538336, -81.379234]
dist = calculate_distance(ksc_coord[0], ksc_coord[1], orlando_coord[0], orlando_coord[1])
print(f"Distance KSC to Orlando: {dist:.2f} km")
```
