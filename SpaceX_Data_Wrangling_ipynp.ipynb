```python
import pandas as pd

print("LOADING...")
url = 'https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBM-DS0321EN-SkillsNetwork/datasets/spacex_launch_dash.csv'
df = pd.read_csv(url)

print("\nRENAMING COLUMNS TO REMOVE SPACES...")
# Rename columns so we don't get KeyErrors
df = df.rename(columns={
    'Flight Number': 'FlightNumber',
    'Launch Site': 'LaunchSite',
    'Payload Mass (kg)': 'PayloadMass',
    'class': 'Class'
})

print("Columns now:", df.columns.tolist())

print("\nSELECTING 4 COLUMNS...")
df_final = df[['FlightNumber', 'PayloadMass', 'LaunchSite', 'Class']]

print("FILLING MISSING VALUES...")
df_final['PayloadMass'] = df_final['PayloadMass'].fillna(df_final['PayloadMass'].mean())

print("SAVING...")
df_final.to_csv('dataset_part1.csv', index=False)

print("\n" + "="*50)
print("✅ SUCCESS!!! DATA WRANGLING COMPLETE")
print("="*50)
print(df_final.head(10))
```
