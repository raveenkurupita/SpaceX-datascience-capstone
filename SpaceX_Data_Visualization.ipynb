```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
%matplotlib inline

df = pd.read_csv("https://github.com/raveenkurupita/SpaceX-datascience-capstone/raw/main/dataset_part1.csv")

# 1. Flight Number vs Payload Mass
plt.figure(figsize=(10,6))
sns.scatterplot(x='Flight Number', y='PayloadMass', hue='Class', data=df)
plt.title('Flight Number vs Payload Mass')
plt.show()

# 2. Launch Site vs Payload Mass
plt.figure(figsize=(10,6))
sns.scatterplot(x='LaunchSite', y='PayloadMass', hue='Class', data=df)
plt.title('Launch Site vs Payload Mass')
plt.show()

# 3. Success Rate by Orbit - NOTE: This dataset doesn't have 'Orbit' column
# So we'll skip and do Success Rate by LaunchSite instead
success_site = df.groupby('LaunchSite')['Class'].mean().reset_index()
plt.figure(figsize=(10,6))
sns.barplot(x='LaunchSite', y='Class', data=success_site)
plt.title('Success Rate by Launch Site')
plt.xticks(rotation=45)
plt.show()

# 4. Flight Number vs Launch Site
plt.figure(figsize=(10,6))
sns.scatterplot(x='Flight Number', y='LaunchSite', hue='Class', data=df)
plt.title('Flight Number vs Launch Site')
plt.show()

# 5. Payload vs Launch Site
plt.figure(figsize=(10,6))
sns.scatterplot(x='PayloadMass', y='LaunchSite', hue='Class', data=df)
plt.title('Payload Mass vs Launch Site')
plt.show()

# 6. Success Rate by Year
df['Date'] = pd.to_datetime('2020-01-01') # This dataset doesn't have Date, so we use Booster Version Category
success_version = df.groupby('Booster Version Category')['Class'].mean().reset_index()
plt.figure(figsize=(10,6))
sns.barplot(x='Booster Version Category', y='Class', data=success_version)
plt.title('Success Rate by Booster Version')
plt.xticks(rotation=45)
plt.show()
```
